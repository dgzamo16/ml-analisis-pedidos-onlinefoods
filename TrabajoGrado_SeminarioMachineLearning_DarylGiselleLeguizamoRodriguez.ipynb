6. Procedemos a utilizar la tecnica de "Aprendizaje No supervisado (Agrupacion)" para desarrollar el analisis de las variables.


- eliminamos datos duplicados para un resultado mas detallado.

```python
# Identificar y eliminar duplicados
print('Número de filas antes de eliminar duplicados:', len(Datos_onlinefoods))
Datos_onlinefoods = Datos_onlinefoods.drop_duplicates()
print('Número de filas después de eliminar duplicados:', len(Datos_onlinefoods))
```

- usamos la tecnica de agrupacion para ver las relaciones entre los datos. Ademas, implementamos estrategias de Clustering para la segmetacion de los datos.

```python
# Importando las bibliotecas necesarias
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Transformando el dataset a formato numérico
Datos_onlinefoods_numerico = np.array(Datos_onlinefoods)

# Elegir el número de grupos
k = 3

#Generando y entrenando el momdelo
Modelo_Cluster = KMeans(n_clusters=k, n_init = 'auto', random_state=42)
Modelo_Cluster.fit(Datos_onlinefoods_numerico)

#Prediciendo etiquetas de grupos en el conjunto de datos
Prediccion = Modelo_Cluster.fit_predict(Datos_onlinefoods_numerico)

# Asignando etiqueta como columna nueva al dataset
Datos_onlinefoods_2 = Datos_onlinefoods.copy()
Datos_onlinefoods_2['Grupos'] = Prediccion
Datos_onlinefoods_2.head(15)
```

```python
#Cargando librerías
import seaborn as sns
import matplotlib.pyplot as plt

# Crear un gráfico de dispersión con Seaborn
sns.scatterplot(data = Datos_onlinefoods_2, x='Monthly Income', y='Family size')
plt.title('Gráfico de Dispersión del salario mensual comparado con el tamaño de la familia')
plt.xlabel('Monthly Income')
plt.ylabel('Family size')
plt.show()
```

- obtenemos los grupos

```python
# Obtener los centroides del modelo K-Means
Grupos=Modelo_Cluster.cluster_centers_

# Obtener los nombres de las columnas del dataset original
nombres_variables = Datos_onlinefoods.columns[:]  # Excluir la última columna que contiene los clusters

# Crear un DataFrame para los centroides con los nombres de las columnas
Grupos_df = pd.DataFrame(Grupos[:,:] , columns=nombres_variables)

# Imprimir los centroides con los nombres de las variables
print("Grupo:")
print(Grupos_df)
```

- Guardamos el nuevo dataset que contiene el procedimiento o resultado de los algoritmos y el modelo entrenado que utilizamos.

```python
# Guardar el dataset con la columna de etiqueta como CSV
Datos_onlinefoods_2.to_csv('OnlineFoods_etiquetas.csv', index=False)

# Guardar los centroides como CSV
Grupos_df.to_csv('Grupos_OnlineFoods.csv', index=False)
```
